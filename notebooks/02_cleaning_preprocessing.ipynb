```python
categorical_cols = [
    'job_title',
    'company_type',
    'industry',
    'country',
    'city',
    'remote_type',
    'experience_level',
    'employment_type',
    'company_size'
]

for col in categorical_cols:
    df[col] = df[col].astype('category')

print("Categorical datatypes fixed successfully")
```

## Outlier Detection Using IQR

Outliers were detected using the Interquartile Range (IQR) method.

The following numerical columns were analyzed:

- min_experience_years
- salary_min_usd
- salary_max_usd
- posted_year

```python
numerical_cols = [
    'min_experience_years',
    'salary_min_usd',
    'salary_max_usd',
    'posted_year'
]

numerical_cols
```

```python
for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"{col}")
    print(f"Outliers: {len(outliers)}")
    print("-" * 40)
```

## Data Transformation

The dataset was evaluated for potential transformations.

Therefore, no additional transformations were required and the dataset was retained in its original cleaned form.
